```python
# 老師提供的載入工具，直接執行，不用修改。
# 執行下一格時會跳出上傳按鈕，請選你從創課下載的 W09_資料.zip。
# 重新啟動後檔案會消失，再上傳一次同一個 zip 即可——這是正常的。
import pathlib
import zipfile


def _find(pattern, roots=(".", "/content")):
    """在幾個資料夾裡找符合的檔案，回傳第一個。"""
    for root in roots:
        base = pathlib.Path(root)
        if not base.exists():
            continue
        hits = sorted(base.rglob(pattern))
        if hits:
            return hits[0]
    return None


def _unzip_all(roots=(".", "/content")):
    """把找得到的 zip 全部解開（解到 zip 所在的資料夾）。"""
    done = []
    for root in roots:
        base = pathlib.Path(root)
        if not base.exists():
            continue
        for z in sorted(base.glob("*.zip")):
            try:
                with zipfile.ZipFile(z) as zf:
                    zf.extractall(z.parent)
                done.append(z.name)
            except Exception as e:
                print("解壓縮失敗：", z.name, str(e)[:60])
    return done


def load_data(filename, quiet=False):
    """取得資料檔的路徑。找不到就請學生上傳當週的 zip。

    filename 可以是完整檔名，也可以是樣式（例如 "cgm_*.csv"）。
    """
    pattern = filename if any(ch in filename for ch in "*?") else "**/" + filename

    hit = _find(pattern)
    if hit:
        if not quiet:
            print("已找到資料：", hit.name)
        return str(hit)

    unzipped = _unzip_all()
    if unzipped:
        hit = _find(pattern)
        if hit:
            if not quiet:
                print("已從", "、".join(unzipped), "解出：", hit.name)
            return str(hit)

    try:
        from google.colab import files  # type: ignore
    except ImportError:
        raise FileNotFoundError(
            f"找不到 {filename}。請把當週的 zip 放在這個資料夾後重新執行。")

    print("請選擇你從創課下載的當週 zip 檔（例如 W05_資料.zip）")
    files.upload()
    _unzip_all()
    hit = _find(pattern)
    if hit:
        if not quiet:
            print("已載入：", hit.name)
        return str(hit)

    raise FileNotFoundError(
        f"上傳完成，但 zip 裡找不到 {filename}。\n"
        "請確認下載的是當週的 zip，或重新從創課下載一次。")

print("載入工具已就緒")
```

```python
RESULTS = {}
EXPECTED = 15          # 本週共有幾項檢查

def check(name, ok):
    # ok 為真才算通過。這個函式只記錄，不會讓你的程式中斷。
    RESULTS[name] = bool(ok)
    print(("[通過] " if ok else "[未通過] ") + name)

def 總表(標題="檢查總表", 應有=None):
    # 應有：這個時點應該跑完幾項。課堂檢查點只算課堂範圍，總表算全部。
    應有 = EXPECTED if 應有 is None else 應有
    passed = sum(1 for v in RESULTS.values() if v)
    print(f"{標題}：通過 {passed} / {應有}")

    missing = 應有 - len(RESULTS)
    if missing > 0:
        print()
        print(f"[注意] 有 {missing} 項檢查沒有跑到。")
        print("  那一格多半出錯中斷了，或是被你註解掉。")
        print("  往上找紅色錯誤訊息的那一格，修好再從頭重跑一次。")
        print("  註解掉不算完成——分母不會變小，只會算你沒通過。")

    failed = [k for k, v in RESULTS.items() if not v]
    if failed:
        print()
        print("未通過的項目：")
        for k in failed:
            print("   ·", k)
        print()
        print("常見原因，照順序檢查：")
        print("  1. 變數還是 None —— TODO 還沒填。找到那一格，把 None 換成你的答案。")
        print("  2. 把出錯的程式註解掉了 —— 註解掉不算完成，錯誤要修好。")
        print("  3. def 寫了但沒有被呼叫 —— 函式光定義不會執行，要有人呼叫它。")
        print("  4. 改完沒有重跑 —— 改了上面的格子，下面的檢查要重新執行才會更新。")
        print("     最保險：「執行階段」→「重新啟動工作階段並執行所有儲存格」。")

    if missing == 0 and not failed:
        print()
        print("全部通過。")

print("檢查工具已載入。本週共", EXPECTED, "項檢查。")
```
